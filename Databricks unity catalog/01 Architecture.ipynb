### Azure Databricks
Azure Databricks is a unified, open analytics platform for building, deploying, sharing, and maintaining enterprise-grade data, analytics, and AI solutions at scale. The Databricks Data + AI Platform integrates with cloud storage and security in your cloud account, and manages and deploys cloud infrastructure for you.

### Data Lakehouse
A data lakehouse is a data management system that combines the benefits of data lakes and data warehouses.A data lakehouse provides scalable storage and processing capabilities for modern organizations that want to avoid isolated systems for processing different workloads, like machine learning (ML) and business intelligence (BI). A data lakehouse can help establish a single source of truth, eliminate redundant costs, and ensure data freshness.

Data lakehouses often use a data design pattern that incrementally improves, enriches, and refines data as it moves through layers of staging and transformation. Each layer of the lakehouse can include one or more layers. This pattern is frequently referred to as a medallion architecture.

The Databricks lakehouse uses two additional key technologies:

**Delta Lake**: an optimized storage layer that supports ACID transactions and schema enforcement.
**Unity Catalog**: a unified, fine-grained governance solution for data and AI.

### Lakehouse vs Data Lake vs Data Warehouse

**Data Warehouse**: Optimizes queries for business intelligence (BI) reports using clean, structured data. They are designed for data that changes infrequently to prevent concurrent query conflicts. However, they often rely on proprietary formats that limit machine learning support, and generating results can take minutes or even hours.

**Data Lake**: Stores and processes raw data of any nature and format cheaply and permanently. While widely used for data science and machine learning, their unvalidated nature makes them unsuitable for reliable BI reporting.

**Data Lakehouse**: Combines the strengths of both by applying an optimized metadata layer over validated data stored in standard cloud object storage. This unified platform allows you to work from the same data across different use cases, providing:
- Open, direct access to standard data formats.
- Indexing protocols optimized for machine learning and data science.
- Low query latency and high reliability for both BI and advanced analytics.

## What is Databricks?
**Databricks** is a management layer built on top of Apache Spark clusters. It automates and manages the underlying clusters (groups of compute nodes) to eliminate the overhead of manual infrastructure management.

## Databricks Architecture
The Databricks architecture is divided into two primary components:

*   **Control Plane:** The frontend management layer. It handles the user interface (UI), web application, access control, cluster configuration, and notebook creation.
*   **Compute Plane:** The backend operational layer. It manages the actual computing resources (clusters, virtual machines, nodes) and handles the data residing in cloud storage.

## Azure Basic Concepts
*   **Resource Group:** A parent folder used to store and organize various cloud resources. 
*   **Resource:** A specific service provisioned from Azure (e.g., Azure Databricks, Azure Data Factory, Storage Account). Every resource must be stored within a Resource Group.

## Storage Account vs. Data Lake
*   **Storage Account (Blob Storage):** Azure's default storage service. Standard Blob storage stores data in flat containers and does not support hierarchical folder structures.
*   **Data Lake:** A storage architecture built on top of Blob storage. In Azure, a Data Lake is created by enabling a **Hierarchical Namespace** within a Storage Account, which allows for the creation of nested, hierarchical folders inside containers.

## Managed Resource Group
A secondary, automatically generated Resource Group controlled entirely by Databricks. It serves as the Compute Plane for the workspace, automatically storing the virtual machines for your clusters and the default storage account for managed tables.


# Azure Databricks Components

This document outlines the fundamental components used to organize, manage, and process data and AI assets in Azure Databricks.

## Accounts and Workspaces
* **Workspace:** A cloud deployment that functions as an environment for accessing all your Databricks assets. It organizes objects (like notebooks and libraries) into folders and provides access to data and compute resources.
* **Account:** A single entity that can include multiple workspaces. Accounts enabled for Unity Catalog allow centralized management of users and data access across all workspaces.

## Billing
* **Databricks Units (DBUs):** Units of processing capability billed per hour, based on the VM instance type used.

## Authentication and Authorization
* **User:** A unique individual with system access, represented by an email address.
* **Service Principal:** A service identity used for automated tools, jobs, scripts, and CI/CD platforms, represented by an application ID.
* **Group:** A collection of identities (users or service principals) that simplifies identity management and access assignment.
* **Access Control List (ACL):** A list of permissions attached to an object (workspace, cluster, table, etc.) specifying which users/processes have access and what operations they can perform.
* **Personal Access Token (PAT):** A string used to authenticate REST API calls and external tool connections.

## Interfaces
* **UI:** The graphical web interface for interacting with workspace folders, data objects, and compute resources.
* **Genie One:** A simplified interface for business users to view AI/BI dashboards and ask natural language data questions without navigating technical concepts.
* **REST API:** Endpoints for modifying or requesting information about account and workspace objects.
* **SQL REST API:** Endpoints specifically designed to automate tasks on SQL objects.
* **CLI:** The Databricks command-line interface, built on top of the REST API and hosted on GitHub.

## Data Management
* **Unity Catalog:** A unified governance solution providing centralized access control, auditing, lineage, and data discovery across workspaces.
* **Catalog:** The highest-level container for organizing and isolating data. Catalogs can be shared across workspaces in the same region and account.
* **Schema (Database):** Containers within catalogs that provide granular organization for tables, volumes, models, and functions.
* **Table:** An object that organizes and governs access to structured data, queryable via Apache Spark SQL and APIs.
* **Delta Tables:** The default table format in Databricks. Based on the open-source Delta Lake, it provides high-performance ACID table storage over cloud object stores.
* **View:** A read-only object derived from one or more tables/views that saves defined queries.
* **Volume:** A logical volume representing cloud object storage used to organize and govern access to non-tabular (unstructured/semi-structured) data.
* **Metastore:** Registers metadata about data, AI, and permissions. Unity Catalog uses an account-level metastore, while a legacy Hive metastore exists for workspaces not using Unity Catalog.
* **Catalog Explorer:** A UI tool to explore, manage, and track lineage for data and AI assets.
* **DBFS Root:** A deprecated, default storage location available to all users. Databricks now recommends Unity Catalog volumes instead.

## Computation Management
* **Cluster:** A set of compute resources and configurations used to run workloads.
  * **All-purpose cluster:** Created manually via UI/CLI/API for collaborative, interactive analysis. Can be restarted.
  * **Job cluster:** Created automatically by the job scheduler to run a specific job and terminated upon completion. Cannot be restarted.
* **Pool:** A set of idle, ready-to-use instances that significantly reduce cluster start and auto-scaling times.
* **Databricks Runtime:** The core component set running on clusters, including Apache Spark and optimizations for big data analytics performance and security.
* **Workload:** The processing capability needed for tasks. Data engineering workloads run on automated job clusters, while data analytics (interactive) workloads run on all-purpose clusters.
* **Execution Context:** The state for a read–eval–print loop (REPL) environment supporting Python, R, Scala, and SQL.

## Orchestration & Data Engineering Tools
* **Jobs:** A non-interactive mechanism for scheduling and orchestrating notebooks, libraries, and pipelines.
* **Pipelines:** Declarative data processing pipelines built on Apache Spark Declarative Pipelines (SDP) for reliability and maintainability.
* **Jobs & Pipelines UI:** The interface serving as the entry point for Jobs, Lakeflow pipelines, and Lakeflow Connect.
* **Notebook:** A web-based interface containing runnable code, narrative text, and visualizations for developing data and ML workflows.
* **Library:** Code packages made available to a notebook or job running on a cluster.
* **Git Folder (formerly Repos):** Workspace folders synced to remote Git repositories to provide source and version control for your data engineering projects.

## AI and Machine Learning
* **Machine Learning Runtime:** A specialized Databricks Runtime automating compute creation with pre-built ML/DL infrastructure (TensorFlow, PyTorch, XGBoost) and GPU support.
* **Experiment:** A collection of MLflow runs used to track the development of agents, LLMs, and ML models.
* **Features:** A feature store ensuring the same feature computation code is used across model training and inference.
* **AI Models:** Environments and capabilities for deploying AI, including an AI playground for prompting LLMs, built-in foundation models, external models, and fine-tuning tools.
* **Model Registry:** A hosted version of the MLflow Model Registry within Unity Catalog for managing the model lifecycle.
* **Model Serving:** A unified interface exposing deployed models as REST APIs for integration into client applications.

## Data Warehousing
* **SQL Warehouse:** A specialized compute resource optimized for running SQL queries (available in Classic, Pro, and Serverless types).
* **Query:** A valid SQL statement authored in the SQL editor or connected via a SQL connector/API.
* **Query History:** A log of executed queries and their performance metrics used to identify bottlenecks.
* **Visualization:** Graphical presentations of query results.
* **Dashboard:** Presentations combining data visualizations and commentary, often auto-distributed for business reporting.
